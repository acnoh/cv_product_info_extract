```python
import easyocr
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = "images/val/IMG_4664.jpeg"
```

```python
reader = easyocr.Reader(['en']) # EasyOCR
ocr_results = reader.readtext(img) # perform OCR on the product image

# print out the OCR results: each result contains a bounding box, the detected text, and a confidence score.
for (bbox, text, confidence) in ocr_results:
    print(f"Detected text: {text} (confidence: {confidence:.2f})")
```

```python
for bbox, text, conf in ocr_results:
    if "net wt." in text.lower() and conf >= 0.75:
        print("Product Size =", text)
```
